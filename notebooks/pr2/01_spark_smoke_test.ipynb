# Phase 2 - Spark Cluster Smoke Test

This notebook verifies that Jupyter can submit work to the Spark master and worker. The same shared session helper is used by the production jobs in `spark_jobs/`.

```python
import sys
from pathlib import Path
import pyspark

PROJECT_ROOT = Path("/workspace")
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from spark_jobs.common import create_spark_session, load_settings

settings = load_settings()
spark = create_spark_session("phase2-spark-smoke-test", settings)
print("PySpark version:", pyspark.__version__)
print("Spark master:", spark.sparkContext.master)
print("Application ID:", spark.sparkContext.applicationId)
```

```python
demo_df = spark.createDataFrame(
    [("JFK", 25.5), ("LAX", 18.2), ("ORD", 30.1)],
    ["airport", "wind_speed_kts"],
)
demo_df.show()
demo_df.groupBy().avg("wind_speed_kts").show()
spark.stop()
print("Spark cluster smoke test complete.")
```
